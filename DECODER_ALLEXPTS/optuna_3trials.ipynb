```python
# =========================
# 0. INSTALL
# =========================
!pip install -q transformers datasets accelerate peft trl bitsandbytes scikit-learn pandas matplotlib seaborn sentencepiece optuna

# =========================
# 1. IMPORTS
# =========================
import os
import re
import gc
import json
import math
import torch
import optuna
import numpy as np
import pandas as pd
from getpass import getpass

from datasets import Dataset
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

from huggingface_hub import login, whoami
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    TrainingArguments,
    BitsAndBytesConfig,
    Trainer,
    DataCollatorForSeq2Seq,
)
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training, PeftModel

# =========================
# 2. AUTH + DRIVE
# =========================
HF_TOKEN = getpass("Enter your Hugging Face token: ").strip()
login(HF_TOKEN)
print("HF user:", whoami())

from google.colab import drive
drive.mount('/content/drive')

# =========================
# 3. CONFIG
# =========================
DATA_PATH = "/content/drive/MyDrive/conversation_en.json"
AUGMENTED_TRAIN_PATH = "/content/drive/MyDrive/final_train_augmented_v2.csv"
MODEL_NAME = "meta-llama/Llama-3.2-3B-Instruct"

OPTUNA_DB_PATH = "/content/drive/MyDrive/llama32_3b_augmented_v2_optuna.db"
OPTUNA_STUDY_NAME = "llama32_3b_augmented_v2_optuna"
OPTUNA_TRIALS_CSV = "/content/drive/MyDrive/llama32_3b_augmented_v2_optuna_trials.csv"
BEST_PARAMS_JSON = "/content/drive/MyDrive/llama32_3b_augmented_v2_best_params.json"

FINAL_OUTPUT_DIR = "/content/drive/MyDrive/llama32_3b_newsft_optuna_best"
FINAL_PRED_PATH = "/content/drive/MyDrive/llama32_3b_optuna_best_test_predictions.csv"

RANDOM_STATE = 42
TEST_SIZE = 0.20
VAL_SIZE_WITHIN_TRAIN = 0.20

VALID_LABELS = ["Chitchat", "Interpret", "Reveal", "Revise", "Inquire"]
MAX_SEQ_LEN = 512

# T4-safe
PER_DEVICE_TRAIN_BATCH_SIZE = 2
PER_DEVICE_EVAL_BATCH_SIZE = 2
GRADIENT_ACCUMULATION_STEPS = 4

# set how many trials to run this session
N_TRIALS_THIS_RUN = 3

os.makedirs(FINAL_OUTPUT_DIR, exist_ok=True)

print("CUDA available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
print("Model:", MODEL_NAME)

# =========================
# 4. LOAD ORIGINAL DATA
# =========================
with open(DATA_PATH, "r", encoding="utf-8") as f:
    raw = json.load(f)

all_rows = []
for conv in raw:
    conv_id = str(conv.get("conv_id"))
    for i, t in enumerate(conv.get("conversation", [])):
        all_rows.append({
            "conversation_id": conv_id,
            "turn_index": int(i),
            "role": str(t.get("role", "")).lower(),
            "text": t.get("content") or "",
            "intent": t.get("intent"),
        })

df_turns_full = (
    pd.DataFrame(all_rows)
    .sort_values(["conversation_id", "turn_index"])
    .reset_index(drop=True)
)

df_turns_full["conversation_id"] = df_turns_full["conversation_id"].astype(str)
df_turns_full["turn_index"] = df_turns_full["turn_index"].astype(int)
df_turns_full["role"] = df_turns_full["role"].astype(str).str.lower()
df_turns_full["text"] = df_turns_full["text"].fillna("").astype(str)

target_df = df_turns_full[
    (df_turns_full["role"] == "user") &
    (df_turns_full["intent"].notna()) &
    (df_turns_full["intent"].isin(VALID_LABELS))
].copy().reset_index(drop=True)

print("Full rows:", len(df_turns_full))
print("Target rows:", len(target_df))
print("Unique conversations:", df_turns_full["conversation_id"].nunique())

# =========================
# 5. SPLIT ORIGINAL DATA
# =========================
gss1 = GroupShuffleSplit(n_splits=1, test_size=TEST_SIZE, random_state=RANDOM_STATE)
train_val_idx, test_idx = next(
    gss1.split(target_df, groups=target_df["conversation_id"])
)

train_val_df = target_df.iloc[train_val_idx].reset_index(drop=True)
test_df = target_df.iloc[test_idx].reset_index(drop=True)

gss2 = GroupShuffleSplit(n_splits=1, test_size=VAL_SIZE_WITHIN_TRAIN, random_state=RANDOM_STATE)
train_idx, val_idx = next(
    gss2.split(train_val_df, groups=train_val_df["conversation_id"])
)

train_df_original = train_val_df.iloc[train_idx].reset_index(drop=True)
val_df = train_val_df.iloc[val_idx].reset_index(drop=True)

turns_by_conv_full = {
    cid: g.sort_values("turn_index").reset_index(drop=True)
    for cid, g in df_turns_full.groupby("conversation_id")
}

print("Original train rows:", len(train_df_original))
print("Val rows           :", len(val_df))
print("Test rows          :", len(test_df))
print("Overlap train-test :", len(set(train_df_original["conversation_id"]) & set(test_df["conversation_id"])))
print("Overlap train-val  :", len(set(train_df_original["conversation_id"]) & set(val_df["conversation_id"])))

# =========================
# 5B. LOAD AUGMENTED TRAIN DATA
# =========================
train_df = pd.read_csv(AUGMENTED_TRAIN_PATH)

train_df["conversation_id"] = train_df["conversation_id"].astype(str)
train_df["turn_index"] = train_df["turn_index"].astype(int)
train_df["role"] = train_df["role"].astype(str).str.lower()
train_df["text"] = train_df["text"].fillna("").astype(str)
train_df["intent"] = train_df["intent"].astype(str)

print("Augmented train rows loaded:", len(train_df))
print(train_df["intent"].value_counts())

if "is_augmented" in train_df.columns:
    print("\nAugmented flag counts:")
    print(train_df["is_augmented"].value_counts())

# =========================
# 6. BUILD CONTEXT
# =========================
def build_history_and_current_turn(
    full_conv: pd.DataFrame,
    current_turn_index: int,
    current_turn_text_override: str = None
):
    history_df = full_conv[full_conv["turn_index"] < current_turn_index].copy()

    history_lines = []
    for _, row in history_df.iterrows():
        role = str(row["role"]).upper().strip()
        text = str(row["text"]).strip()
        if text:
            history_lines.append(f"{role}: {text}")

    history_text = "\n".join(history_lines)

    if current_turn_text_override is not None:
        current_turn_text = str(current_turn_text_override).strip()
    else:
        current_df = full_conv[
            (full_conv["turn_index"] == current_turn_index) &
            (full_conv["role"] == "user")
        ].copy()
        current_turn_text = str(current_df.iloc[0]["text"]).strip() if len(current_df) > 0 else ""

    return history_text, current_turn_text


def build_prompt(history_text: str, current_turn_text: str) -> str:
    return f"""### Task:
You are an e-commerce conversational AI intent classifier.

Predict the intent of the current user utterance using the available dialogue context.

### Intent Labels:
- Chitchat: casual or social talk not directly contributing new shopping constraints
- Reveal: the user starts or reveals a shopping need or request
- Interpret: the user provides information, answers a question, or adds preferences/details
- Revise: the user modifies, corrects, or refines a previous request or suggestion
- Inquire: the user asks for additional information, clarification, or details

### Dialogue Context:
{history_text if history_text.strip() else "NONE"}

### Current User Utterance:
{current_turn_text}

### Instruction:
Return exactly one label from:
Chitchat, Reveal, Interpret, Revise, Inquire

### Answer:"""


def build_completion(label: str) -> str:
    return f" {label}"


def build_inference_prompt(history_text: str, current_turn_text: str) -> str:
    return f"""### Task:
You are an e-commerce conversational AI intent classifier.

Predict the intent of the current user utterance using the available dialogue context.

### Intent Labels:
- Chitchat: casual or social talk not directly contributing new shopping constraints
- Reveal: the user starts or reveals a shopping need or request
- Interpret: the user provides information, answers a question, or adds preferences/details
- Revise: the user modifies, corrects, or refines a previous request or suggestion
- Inquire: the user asks for additional information, clarification, or details

### Dialogue Context:
{history_text if history_text.strip() else "NONE"}

### Current User Utterance:
{current_turn_text}

### Instruction:
Return exactly one label from:
Chitchat, Reveal, Interpret, Revise, Inquire

### Answer:"""


def build_sft_dataframe(df_split: pd.DataFrame, turns_by_conv: dict):
    rows = []

    for _, row in df_split.iterrows():
        conv_df = turns_by_conv[row["conversation_id"]]

        history_text, current_turn_text = build_history_and_current_turn(
            conv_df,
            row["turn_index"],
            current_turn_text_override=row["text"]
        )

        prompt = build_prompt(history_text, current_turn_text)
        completion = build_completion(row["intent"])

        rows.append({
            "conversation_id": row["conversation_id"],
            "turn_index": row["turn_index"],
            "history_text": history_text,
            "current_turn_text": current_turn_text,
            "intent": row["intent"],
            "prompt": prompt,
            "completion": completion,
            "text": prompt + completion
        })

    return pd.DataFrame(rows)

train_sft_df = build_sft_dataframe(train_df, turns_by_conv_full)
val_sft_df   = build_sft_dataframe(val_df, turns_by_conv_full)
test_sft_df  = build_sft_dataframe(test_df, turns_by_conv_full)

print("Train SFT size:", len(train_sft_df))
print("Val SFT size  :", len(val_sft_df))
print("Test SFT size :", len(test_sft_df))

# =========================
# 7. LABEL NORMALIZATION
# =========================
def normalize_label(text: str) -> str:
    text = (text or "").strip()
    if not text:
        return "UNKNOWN"

    text = text.split("\n")[0].strip()
    text = re.sub(r"^[\"'`]+|[\"'`]+$", "", text)
    text = re.sub(r"[.,:;!?]+$", "", text).strip()

    label_map = {label.lower(): label for label in VALID_LABELS}
    lowered = text.lower()

    if lowered in label_map:
        return label_map[lowered]

    patterns = [
        r"\b(chitchat|reveal|interpret|revise|inquire)\b",
        r"label\s*[:\-]?\s*(chitchat|reveal|interpret|revise|inquire)\b",
        r"intent\s*[:\-]?\s*(chitchat|reveal|interpret|revise|inquire)\b",
        r"the answer is\s*(chitchat|reveal|interpret|revise|inquire)\b",
        r"the intent is\s*(chitchat|reveal|interpret|revise|inquire)\b",
        r"answer\s*[:\-]?\s*(chitchat|reveal|interpret|revise|inquire)\b",
    ]

    for pattern in patterns:
        match = re.search(pattern, lowered, flags=re.IGNORECASE)
        if match:
            return label_map[match.group(1).lower()]

    return "UNKNOWN"

# =========================
# 8. TOKENIZER
# =========================
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, trust_remote_code=True)
if tokenizer.pad_token_id is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"

# =========================
# 9. LENGTH CHECK (OPTIONAL)
# =========================
def get_total_len(prompt, completion):
    prompt_ids = tokenizer(prompt, add_special_tokens=False)["input_ids"]
    completion_ids = tokenizer(completion, add_special_tokens=False)["input_ids"]
    if tokenizer.eos_token_id is not None:
        completion_ids = completion_ids + [tokenizer.eos_token_id]
    return len(prompt_ids) + len(completion_ids)

lengths_train = [get_total_len(r["prompt"], r["completion"]) for _, r in train_sft_df.iterrows()]
lengths_val   = [get_total_len(r["prompt"], r["completion"]) for _, r in val_sft_df.iterrows()]
lengths_test  = [get_total_len(r["prompt"], r["completion"]) for _, r in test_sft_df.iterrows()]

print("\nLength summary")
print("Train max:", max(lengths_train))
print("Val max  :", max(lengths_val))
print("Test max :", max(lengths_test))

# =========================
# 10. OBJECTIVE FUNCTION
# =========================
def objective(trial):
    gc.collect()
    torch.cuda.empty_cache()

    learning_rate = trial.suggest_float("learning_rate", 1e-5, 3e-4, log=True)
    num_train_epochs = trial.suggest_categorical("num_train_epochs", [2, 3, 4])
    warmup_ratio = trial.suggest_float("warmup_ratio", 0.0, 0.1)
    weight_decay = trial.suggest_float("weight_decay", 0.0, 0.1)
    lora_r = trial.suggest_categorical("lora_r", [8, 16])
    lora_alpha = trial.suggest_categorical("lora_alpha", [16, 32])
    lora_dropout = trial.suggest_float("lora_dropout", 0.0, 0.1)

    print(f"\n=== Trial {trial.number} ===")
    print({
        "learning_rate": learning_rate,
        "num_train_epochs": num_train_epochs,
        "warmup_ratio": warmup_ratio,
        "weight_decay": weight_decay,
        "lora_r": lora_r,
        "lora_alpha": lora_alpha,
        "lora_dropout": lora_dropout,
        "max_seq_len": MAX_SEQ_LEN,
    })

    train_ds = Dataset.from_pandas(
        train_sft_df[["prompt", "completion"]],
        preserve_index=False
    )
    val_ds = Dataset.from_pandas(
        val_sft_df[["prompt", "completion"]],
        preserve_index=False
    )

    def tokenize_response_only(example):
        prompt_ids = tokenizer(
            example["prompt"],
            add_special_tokens=False,
            truncation=False
        )["input_ids"]

        completion_ids = tokenizer(
            example["completion"],
            add_special_tokens=False,
            truncation=False
        )["input_ids"]

        if tokenizer.eos_token_id is not None:
            completion_ids = completion_ids + [tokenizer.eos_token_id]

        input_ids = prompt_ids + completion_ids
        labels = [-100] * len(prompt_ids) + completion_ids
        attention_mask = [1] * len(input_ids)

        if len(input_ids) > MAX_SEQ_LEN:
            overflow = len(input_ids) - MAX_SEQ_LEN
            input_ids = input_ids[overflow:]
            attention_mask = attention_mask[overflow:]
            labels = labels[overflow:]

        return {
            "input_ids": input_ids,
            "attention_mask": attention_mask,
            "labels": labels,
        }

    train_tokenized = train_ds.map(
        tokenize_response_only,
        remove_columns=train_ds.column_names
    )
    val_tokenized = val_ds.map(
        tokenize_response_only,
        remove_columns=val_ds.column_names
    )

    bnb_config = BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_quant_type="nf4",
        bnb_4bit_use_double_quant=True,
        bnb_4bit_compute_dtype=torch.float16,
    )

    base_model = AutoModelForCausalLM.from_pretrained(
        MODEL_NAME,
        quantization_config=bnb_config,
        device_map="auto",
        torch_dtype=torch.float16,
        trust_remote_code=True
    )

    base_model.config.use_cache = False
    base_model = prepare_model_for_kbit_training(base_model)

    lora_config = LoraConfig(
        r=lora_r,
        lora_alpha=lora_alpha,
        lora_dropout=lora_dropout,
        bias="none",
        task_type="CAUSAL_LM",
        target_modules=["q_proj", "k_proj", "v_proj", "o_proj"]
    )

    model = get_peft_model(base_model, lora_config)
    model.gradient_checkpointing_enable()

    for name, param in model.named_parameters():
        if param.requires_grad:
            param.data = param.data.to(torch.float32)

    trial_output_dir = f"/content/drive/MyDrive/llama32_3b_optuna_trial_{trial.number}"

    training_args = TrainingArguments(
        output_dir=trial_output_dir,
        per_device_train_batch_size=PER_DEVICE_TRAIN_BATCH_SIZE,
        per_device_eval_batch_size=PER_DEVICE_EVAL_BATCH_SIZE,
        gradient_accumulation_steps=GRADIENT_ACCUMULATION_STEPS,
        learning_rate=learning_rate,
        num_train_epochs=num_train_epochs,
        logging_steps=20,
        eval_strategy="epoch",
        save_strategy="no",
        load_best_model_at_end=False,
        report_to="none",
        fp16=False,
        bf16=False,
        warmup_ratio=warmup_ratio,
        weight_decay=weight_decay,
        lr_scheduler_type="cosine",
        max_grad_norm=0.0,
    )

    data_collator = DataCollatorForSeq2Seq(
        tokenizer=tokenizer,
        model=model,
        padding=True,
        return_tensors="pt"
    )

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_tokenized,
        eval_dataset=val_tokenized,
        data_collator=data_collator,
    )

    trainer.train()

    # Validate with generation-based metric
    model.eval()
    val_preds = []

    for _, row in val_sft_df.iterrows():
        prompt = build_inference_prompt(row["history_text"], row["current_turn_text"])

        inputs = tokenizer(
            prompt,
            return_tensors="pt",
            truncation=True,
            max_length=MAX_SEQ_LEN
        )
        inputs = {k: v.to(model.device) for k, v in inputs.items()}

        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                max_new_tokens=3,
                do_sample=False,
                temperature=0.0,
                pad_token_id=tokenizer.pad_token_id,
                eos_token_id=tokenizer.eos_token_id
            )

        input_len = inputs["input_ids"].shape[1]
        new_tokens = outputs[0][input_len:]
        raw_output = tokenizer.decode(new_tokens, skip_special_tokens=True).strip()
        pred_label = normalize_label(raw_output)

        val_preds.append(pred_label if pred_label in VALID_LABELS else "UNKNOWN")

    val_report = classification_report(
        val_sft_df["intent"].tolist(),
        val_preds,
        labels=VALID_LABELS + ["UNKNOWN"],
        digits=3,
        zero_division=0,
        output_dict=True
    )

    macro_f1 = val_report["macro avg"]["f1-score"]
    revise_f1 = val_report.get("Revise", {}).get("f1-score", 0.0)
    inquire_f1 = val_report.get("Inquire", {}).get("f1-score", 0.0)

    trial.set_user_attr("macro_f1", macro_f1)
    trial.set_user_attr("revise_f1", revise_f1)
    trial.set_user_attr("inquire_f1", inquire_f1)

    print("\nValidation result:")
    print({
        "macro_f1": macro_f1,
        "revise_f1": revise_f1,
        "inquire_f1": inquire_f1,
    })

    del trainer
    del model
    del base_model
    gc.collect()
    torch.cuda.empty_cache()

    return macro_f1

# =========================
# 11. CREATE / RESUME STUDY
# =========================
study = optuna.create_study(
    study_name=OPTUNA_STUDY_NAME,
    direction="maximize",
    storage=f"sqlite:///{OPTUNA_DB_PATH}",
    load_if_exists=True,
    sampler=optuna.samplers.TPESampler(seed=RANDOM_STATE)
)

print("Existing trials:", len(study.trials))
study.optimize(objective, n_trials=N_TRIALS_THIS_RUN)

print("\nBest trial so far:")
print("Best value :", study.best_value)
print("Best params:", study.best_params)

with open(BEST_PARAMS_JSON, "w") as f:
    json.dump({
        "best_value": study.best_value,
        "best_params": study.best_params,
    }, f, indent=2)

study.trials_dataframe().to_csv(OPTUNA_TRIALS_CSV, index=False)

print("\nSaved study to:")
print(OPTUNA_DB_PATH)
print(OPTUNA_TRIALS_CSV)
print(BEST_PARAMS_JSON)

# =========================
# 12. RETRAIN FINAL MODEL WITH BEST PARAMS
# =========================
best_params = study.best_params
print("\nRetraining final model with best params:")
print(best_params)

gc.collect()
torch.cuda.empty_cache()

train_ds = Dataset.from_pandas(
    train_sft_df[["prompt", "completion"]],
    preserve_index=False
)
val_ds = Dataset.from_pandas(
    val_sft_df[["prompt", "completion"]],
    preserve_index=False
)

def tokenize_response_only(example):
    prompt_ids = tokenizer(
        example["prompt"],
        add_special_tokens=False,
        truncation=False
    )["input_ids"]

    completion_ids = tokenizer(
        example["completion"],
        add_special_tokens=False,
        truncation=False
    )["input_ids"]

    if tokenizer.eos_token_id is not None:
        completion_ids = completion_ids + [tokenizer.eos_token_id]

    input_ids = prompt_ids + completion_ids
    labels = [-100] * len(prompt_ids) + completion_ids
    attention_mask = [1] * len(input_ids)

    if len(input_ids) > MAX_SEQ_LEN:
        overflow = len(input_ids) - MAX_SEQ_LEN
        input_ids = input_ids[overflow:]
        attention_mask = attention_mask[overflow:]
        labels = labels[overflow:]

    return {
        "input_ids": input_ids,
        "attention_mask": attention_mask,
        "labels": labels,
    }

train_tokenized = train_ds.map(
    tokenize_response_only,
    remove_columns=train_ds.column_names
)
val_tokenized = val_ds.map(
    tokenize_response_only,
    remove_columns=val_ds.column_names
)

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.float16,
)

base_model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=bnb_config,
    device_map="auto",
    torch_dtype=torch.float16,
    trust_remote_code=True
)

base_model.config.use_cache = False
base_model = prepare_model_for_kbit_training(base_model)

lora_config = LoraConfig(
    r=best_params["lora_r"],
    lora_alpha=best_params["lora_alpha"],
    lora_dropout=best_params["lora_dropout"],
    bias="none",
    task_type="CAUSAL_LM",
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj"]
)

model = get_peft_model(base_model, lora_config)
model.gradient_checkpointing_enable()

for name, param in model.named_parameters():
    if param.requires_grad:
        param.data = param.data.to(torch.float32)

training_args = TrainingArguments(
    output_dir=FINAL_OUTPUT_DIR,
    per_device_train_batch_size=PER_DEVICE_TRAIN_BATCH_SIZE,
    per_device_eval_batch_size=PER_DEVICE_EVAL_BATCH_SIZE,
    gradient_accumulation_steps=GRADIENT_ACCUMULATION_STEPS,
    learning_rate=best_params["learning_rate"],
    num_train_epochs=best_params["num_train_epochs"],
    logging_steps=20,
    eval_strategy="epoch",
    save_strategy="epoch",
    save_total_limit=2,
    load_best_model_at_end=True,
    report_to="none",
    fp16=False,
    bf16=False,
    warmup_ratio=best_params["warmup_ratio"],
    weight_decay=best_params["weight_decay"],
    lr_scheduler_type="cosine",
    max_grad_norm=0.0,
)

data_collator = DataCollatorForSeq2Seq(
    tokenizer=tokenizer,
    model=model,
    padding=True,
    return_tensors="pt"
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_tokenized,
    eval_dataset=val_tokenized,
    data_collator=data_collator,
)

trainer.train()

trainer.model.save_pretrained(FINAL_OUTPUT_DIR)
tokenizer.save_pretrained(FINAL_OUTPUT_DIR)

print("Saved best Optuna model to:", FINAL_OUTPUT_DIR)

# =========================
# 13. RELOAD FOR TEST EVAL
# =========================
del trainer
gc.collect()
torch.cuda.empty_cache()

tokenizer = AutoTokenizer.from_pretrained(FINAL_OUTPUT_DIR, trust_remote_code=True)
if tokenizer.pad_token_id is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"

base_eval_model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=bnb_config,
    device_map="auto",
    torch_dtype=torch.float16,
    trust_remote_code=True
)

ft_model = PeftModel.from_pretrained(base_eval_model, FINAL_OUTPUT_DIR)
ft_model.eval()

print("Best fine-tuned model reloaded.")

def predict_label_ft(model, tokenizer, history_text: str, current_turn_text: str):
    prompt = build_inference_prompt(history_text, current_turn_text)

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=MAX_SEQ_LEN
    )
    inputs = {k: v.to(model.device) for k, v in inputs.items()}

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=3,
            do_sample=False,
            temperature=0.0,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id
        )

    input_len = inputs["input_ids"].shape[1]
    new_tokens = outputs[0][input_len:]
    raw_output = tokenizer.decode(new_tokens, skip_special_tokens=True).strip()

    pred = normalize_label(raw_output)
    return pred, raw_output

# =========================
# 14. TEST PREDICTION
# =========================
results = []

for _, row in test_sft_df.iterrows():
    pred_label, raw_output = predict_label_ft(
        ft_model,
        tokenizer,
        row["history_text"],
        row["current_turn_text"]
    )

    results.append({
        "conversation_id": row["conversation_id"],
        "turn_index": row["turn_index"],
        "history_text": row["history_text"],
        "current_turn_text": row["current_turn_text"],
        "intent": row["intent"],
        "predicted_label": pred_label,
        "raw_response": raw_output
    })

pred_df = pd.DataFrame(results)
print(pred_df.head())

# =========================
# 15. TEST EVAL
# =========================
unknown_count = (~pred_df["predicted_label"].isin(VALID_LABELS)).sum()
print("Rows with valid predictions:", pred_df["predicted_label"].isin(VALID_LABELS).sum())
print("Rows with UNKNOWN/ERROR predictions:", unknown_count)

eval_df = pred_df.copy()
eval_df["predicted_label_eval"] = eval_df["predicted_label"].where(
    eval_df["predicted_label"].isin(VALID_LABELS),
    "UNKNOWN"
)

labels_for_report = VALID_LABELS + ["UNKNOWN"]

y_true = eval_df["intent"]
y_pred = eval_df["predicted_label_eval"]

report_text = classification_report(
    y_true, y_pred,
    labels=labels_for_report,
    digits=3,
    zero_division=0
)

report_dict = classification_report(
    y_true, y_pred,
    labels=labels_for_report,
    digits=3,
    zero_division=0,
    output_dict=True
)

print(report_text)

cm = confusion_matrix(y_true, y_pred, labels=labels_for_report)
cm_df = pd.DataFrame(
    cm,
    index=[f"true_{x}" for x in labels_for_report],
    columns=[f"pred_{x}" for x in labels_for_report]
)

print(cm_df)

plt.figure(figsize=(8, 7))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=labels_for_report,
    yticklabels=labels_for_report,
    cmap="Blues"
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Llama-3.2-3B Optuna Best Confusion Matrix")
plt.tight_layout()
plt.show()

pred_df.to_csv(FINAL_PRED_PATH, index=False)
print("Saved predictions to:", FINAL_PRED_PATH)

summary_row = {
    "model": MODEL_NAME,
    "setup": "Augmented train v2 + Optuna + Response-only FT + PEFT/LoRA",
    "accuracy": report_dict["accuracy"],
    "macro_f1": report_dict["macro avg"]["f1-score"],
    "weighted_f1": report_dict["weighted avg"]["f1-score"],
    "chitchat_f1": report_dict.get("Chitchat", {}).get("f1-score", 0.0),
    "interpret_f1": report_dict.get("Interpret", {}).get("f1-score", 0.0),
    "reveal_f1": report_dict.get("Reveal", {}).get("f1-score", 0.0),
    "revise_f1": report_dict.get("Revise", {}).get("f1-score", 0.0),
    "inquire_f1": report_dict.get("Inquire", {}).get("f1-score", 0.0),
    "unknown_f1": report_dict.get("UNKNOWN", {}).get("f1-score", 0.0),
    "unknown_count": int(unknown_count)
}

summary_df = pd.DataFrame([summary_row])
print(summary_df)
```
